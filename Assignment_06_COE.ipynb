```python
import pandas as pd

# Sample data
data = {
    'age': [25, 32, 40, 29, 35, None],
    'salary': [50000, 60000, None, 45000, 70000, 65000],
    'gender': ['Male', 'Female', 'Female', 'Male', 'Female', 'Male'],
    'join_date': ['2020-01-15', '2019-06-23', '2021-08-04', '2022-01-10', '2020-03-15', '2019-11-22']
}

df = pd.DataFrame(data)
```

```python
# Fill missing values
df['age'].fillna(df['age'].mean(), inplace=True)  # Fill missing age with the mean age
df['salary'].fillna(df['salary'].median(), inplace=True)  # Fill missing salary with the median salary
```

```python
from sklearn.preprocessing import LabelEncoder

# Convert categorical variable 'gender' to numeric
label_encoder = LabelEncoder()
df['gender_encoded'] = label_encoder.fit_transform(df['gender'])
```

```python
from sklearn.preprocessing import StandardScaler

# Scale features
scaler = StandardScaler()
df[['age', 'salary']] = scaler.fit_transform(df[['age', 'salary']])
```

```python
# Convert 'join_date' to datetime format
df['join_date'] = pd.to_datetime(df['join_date'])

# Extract year, month, and day from 'join_date'
df['join_year'] = df['join_date'].dt.year
df['join_month'] = df['join_date'].dt.month
df['join_day'] = df['join_date'].dt.day
```

```python
# Example: Creating a feature for salary categories
df['salary_category'] = pd.cut(df['salary'], bins=[-float('inf'), -1, 0, 1, float('inf')], labels=['Very Low', 'Low', 'Medium', 'High'])
```

```python
# Binning 'age' into categories
df['age_category'] = pd.cut(df['age'], bins=[0, 20, 30, 40, 50, 100], labels=['<20', '20-30', '30-40', '40-50', '50+'])
```
